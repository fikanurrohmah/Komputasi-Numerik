```python
# Program SPLTV menggunakan Metode Eliminasi Gauss

print("===================================")
print(" PROGRAM PENYELESAIAN SPLTV")
print("       METODE ELIMINASI GAUSS")
print("===================================")
```

```python
# Memasukkan koefisien dan konstanta

persamaan = [
    [2, 3, 1, 17],
    [3, -1, 2, 17],
    [1, 2, 3, 22]
]

print("Persamaan SPLTV:")
print("2x + 3y + z = 17")
print("3x - y + 2z = 17")
print("x + 2y + 3z = 22")
```

```python
# Melakukan proses eliminasi Gauss

for i in range(3):

    if abs(persamaan[i][i]) < 1e-10:
        for j in range(i + 1, 3):
            if abs(persamaan[j][i]) > 1e-10:
                persamaan[i], persamaan[j] = persamaan[j], persamaan[i]
                break

    if abs(persamaan[i][i]) < 1e-10:
        raise ValueError("Persamaan tidak memiliki solusi tunggal.")

    for j in range(i + 1, 3):
        faktor = persamaan[j][i] / persamaan[i][i]

        for k in range(i, 4):
            persamaan[j][k] = persamaan[j][k] - faktor * persamaan[i][k]

print("Proses eliminasi selesai.")
print("Hasil setelah eliminasi:")
print(persamaan)
```

```python
# Menghitung nilai variabel menggunakan substitusi balik

z = persamaan[2][3] / persamaan[2][2]

y = (persamaan[1][3] - persamaan[1][2] * z) / persamaan[1][1]

x = (persamaan[0][3] - persamaan[0][1] * y - persamaan[0][2] * z) / persamaan[0][0]

print("Hasil substitusi balik:")
print("z =", z)
print("y =", y)
print("x =", x)
```

```python
# Menampilkan hasil akhir perhitungan

print("===================================")
print("       HASIL PENYELESAIAN")
print("===================================")

print("Nilai x =", round(x, 2))
print("Nilai y =", round(y, 2))
print("Nilai z =", round(z, 2))

print("===================================")
```
