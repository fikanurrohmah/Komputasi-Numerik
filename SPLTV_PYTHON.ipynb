```python
# Program SPLTV menggunakan Metode Eliminasi Gauss

print("===================================")
print(" PROGRAM PENYELESAIAN SPLTV")
print("       METODE ELIMINASI GAUSS")
print("===================================")
```

```python
# Memasukkan koefisien dan konstanta
# Setiap baris berisi koefisien x, y, z, dan konstanta

persamaan = [
    [2, 3, 1, 17],
    [3, -1, 2, 17],
    [1, 2, 3, 22]
]

# Menampilkan persamaan SPLTV yang akan diselesaikan
print("Persamaan SPLTV:")
print("2x + 3y + z = 17")
print("3x - y + 2z = 17")
print("x + 2y + 3z = 22")
```

```python
# Melakukan proses eliminasi Gauss
# Perulangan dilakukan sebanyak 3 kali untuk setiap variabel

for i in range(3):

     # Memeriksa apakah nilai pembagi mendekati nol
    if abs(persamaan[i][i]) < 1e-10:

       # Mencari baris lain yang memiliki nilai pembagi tidak nol
        for j in range(i + 1, 3):

           # Memeriksa nilai pada baris berikutnya
            if abs(persamaan[j][i]) > 1e-10:

               # Menukar posisi baris jika diperlukan
                persamaan[i], persamaan[j] = persamaan[j], persamaan[i]
                break

    # Memastikan nilai pembagi tidak bernilai nol
    if abs(persamaan[i][i]) < 1e-10:

       # Menampilkan pesan kesalahan jika tidak ditemukan pembagi
        raise ValueError("Persamaan tidak memiliki solusi tunggal.")

    # Menghilangkan variabel pada baris berikutnya
    for j in range(i + 1, 3):

      # Menghitung faktor untuk menghilangkan variabel
        faktor = persamaan[j][i] / persamaan[i][i]

         # Mengubah nilai koefisien dan konstanta
        for k in range(i, 4):
            persamaan[j][k] = persamaan[j][k] - faktor * persamaan[i][k]

# Menampilkan pesan bahwa eliminasi telah selesai
print("Proses eliminasi selesai.")
print("Hasil setelah eliminasi:")
print(persamaan)
```

```python
# Menghitung nilai variabel menggunakan substitusi balik

# Menghitung nilai z dari persamaan terakhir
z = persamaan[2][3] / persamaan[2][2]

# Menghitung nilai y menggunakan nilai z
y = (persamaan[1][3] - persamaan[1][2] * z) / persamaan[1][1]

# Menghitung nilai x menggunakan nilai y dan z
x = (persamaan[0][3] - persamaan[0][1] * y - persamaan[0][2] * z) / persamaan[0][0]

# Menampilkan hasil substitusi balik
print("Hasil substitusi balik:")
print("z =", z)
print("y =", y)
print("x =", x)
```

```python
# Menampilkan hasil akhir perhitungan

print("===================================")
print("       HASIL PENYELESAIAN")
print("===================================")

# Menampilkan nilai x dengan 2 angka di belakang koma
print("Nilai x =", round(x, 2))

# Menampilkan nilai y dengan 2 angka di belakang koma
print("Nilai y =", round(y, 2))

# Menampilkan nilai z dengan 2 angka di belakang koma
print("Nilai z =", round(z, 2))

print("===================================")
```
